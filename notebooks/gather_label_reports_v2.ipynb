```python
def reformat_date(x):
    xsplit = x.split(" ") # remove time if present
    date = xsplit[0]  # keep only the date part
    if "/" in date:
        day, month, year = date.split("/")
        return f"{year}-{month}-{day}"
    else:
        return date
labelled_reports_all["reportDate_orig"] = labelled_reports_all.reportDate
labelled_reports_all.reportDate = labelled_reports_all.reportDate.apply(reformat_date)
labelled_reports_all.reportDate = pd.to_datetime(labelled_reports_all.reportDate, errors="raise")
```

```python
labelled_reports_all[["reportDate_orig", "reportDate"]].groupby("reportDate").head(1)
```

```python
# replace impactSubtype with impactSubType if missing
def replace_impact_subtype(row):
    if pd.isnull(row["impactSubtype"]) and not pd.isnull(row["impactSubType"]):
        row["impactSubtype"] = row["impactSubType"]
    return row
labelled_reports_all = labelled_reports_all.apply(replace_impact_subtype, axis=1)
labelled_reports_all[["impactSubtype", "impactSubType"]].isnull().sum()
```
