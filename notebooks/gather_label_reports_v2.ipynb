```python
def reformat_date(x):
    if "/" in x:
        day, month, year = x.split("/")
        return f"{year}-{month}-{day}"
    else:
        return x
labelled_reports_all["reportDate_orig"] = labelled_reports_all.reportDate
labelled_reports_all.reportDate = labelled_reports_all.reportDate.apply(reformat_date)
labelled_reports_all.reportDate = pd.to_datetime(labelled_reports_all.reportDate, errors="coerce")
```
